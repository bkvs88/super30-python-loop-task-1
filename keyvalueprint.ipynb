### Iterate through student = {"name": "Rahul","age": 22,"course": "Data Science","city": "Bangalore" } and print every key and value.

```python
student = {
"name": "Rahul",
"age": 22,
"course": "Data Science",
"city": "Bangalore"
}
for key,value in student.items():
    print(f"{key} : {value} " )
```
