### Find the largest number from a list without using max().

```python
numbers = [12, 7, 9, 20, 33, 42, 8, 15, 25, 99, 86]
large_number=0
for i in numbers:
    if i>large_number:
        large_number=i
    else:
        large_number=large_number
print(f'{large_number} is the Largest number from the given list {numbers} ')
```
